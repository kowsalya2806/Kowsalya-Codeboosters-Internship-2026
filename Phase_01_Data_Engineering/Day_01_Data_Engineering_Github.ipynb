```python
#step 8: checking the missing values

print("Missing values in each column:")
print(df.isnull().sum())
print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
#step 9: explore specific column

print("Student Names:")
print(df['name'])
print("Student age:")
print(df['age'])
```

```python
#view unique dept

print("Departments in this dataset:")
print(df['department'].unique())

print("\nHow many students per department:")
print(df['department'].value_counts())
```

```python
#step 10: filtering the data

cs_students = df[df['department']=='Computer Science']

print(f"Number of Computer Science students: {len(cs_students)}")

print("\nComputer Science Students:")
print(cs_students[['name','math_score','programming_score','attendance_percentage']])
```

```python
#filter students wit high attendance(above 90)

high_attendance =df[df['attendance_percentage']>90]

print(f"Students with attendance above 90%: {len(high_attendance)}")
print (high_attendance[['name','department','attendance_percentage']])
```

```python
#to find the top student

df['total_score']=df['math_score']+df['science_score']+df['english_score']+df['programming_score']
top_student = df.loc[df['total_score'].idxmax()]
print("Top Students:")
print(f"Name:{top_student['name']}")
print(f"Department:{top_student['department']}")
print(f"Total Score:{top_student['total_score']}")
print(f"Attendance:{top_student['attendance_percentage']}")
```

```python
#calculate avg score per dept

dept_avg = df.groupby('department')['total_score'].mean().round(2)
print("Averge Total Score by Department:")
print(dept_avg)
```

```python
#sort the students by total score

df_sorted = df.sort_values(by='total_score',ascending=False)
print("Top 10 students by total score:")
print(df_sorted[['name','department','total_score','attendance_percentage']].head(10))
```

```python
#how many students from the electronic dept

elect_students = df[df['department']=='Electronics']

print(f"Number of Electronics students: {len(elect_students)}")
```

```python
#AVG attendance of female students

female_st=df[df['gender']=='Female']
avg=female_st['attendance_percentage'].mean()
print('Avg attendance:',avg)
```

```python
#who is the student with the lowest programming score
Low = df[df['programming_score']<40]
print(f"Students lowest score: {len(Low)}")
print(Low[['name','department','attendance_percentage']])
```

```python
#how many students  score above 80 in math

std=df[df['math_score']>80]
print(f"Student who score above 80:{len(std)}")
print(std[['name','department','math_score']])
```
